```python
# Identity Operators
a = 10
b = 20
c = a

print(a is not b)
print(a is c)

# Membership Operators
x = 24
y = 20
my_list = [10, 20, 30, 40, 50]

if x not in my_list:
    print("x is NOT present in given list")
else:
    print("x is present in given list")

if y in my_list:
    print("y is present in given list")
else:
    print("y is NOT present in given list")
```

```python
# Numeric Data Types
a = 5
b = 5.0
c = 2 + 4j

print(type(a))
print(type(b))
print(type(c))

# Sequence Data Type - String
s = 'Welcome to the Geeks World'
print(s)
print(type(s))
print(s[0])   # Access via index
print(s[-1])  # Negative indexing

# Sequence Data Type - List
a = [1, 2, 3]
print(a)

b = ["Geeks", "For", "Geeks", 4, 5]
print(b[0])
print(b[-3])

# Sequence Data Type - Tuple
t1 = (1,)
print(type(t1))

t2 = ('Geeks', 'For', 'Geeks', 1, 2)
print(t2[3])
print(t2[-3])

# Boolean Data Type
print(type(True))
print(type(False))

# Set Data Type
s1 = {"a", "a", "b", "c", "b"}
print(s1)

s2 = ("Geeks", "For", "Geeks")
for i in s2:
    print(i)

# Dictionary Data Type
d = {1: 'Geeks', 2: 'For', 3: 'Geeks'}
print(d[1])
print(d.get(2))
```
